```python
# ============================================
# 04_fixed_smartmeter_dataset_downloader.ipynb
# USING UCI ENERGY DATA (NO ERRORS)
# ============================================

import json
import shutil
import zipfile
from pathlib import Path
import requests
from google.colab import drive, files

# -----------------------------
# 1. Dataset metadata
# -----------------------------
DATASET_NAME = "UCI Appliances Energy Prediction Dataset"
SOURCE_PAGE = "https://archive.ics.uci.edu/ml/datasets/Appliances+energy+prediction"
DOWNLOAD_URL = "https://archive.ics.uci.edu/ml/machine-learning-databases/00374/energydata_complete.csv"

print("Dataset Name :", DATASET_NAME)
print("Source Page  :", SOURCE_PAGE)
print("Download URL :", DOWNLOAD_URL)

# -----------------------------
# 2. Mount Drive
# -----------------------------
drive.mount('/content/drive')

# -----------------------------
# 3. Paths
# -----------------------------
dataset_root = Path("/content/drive/MyDrive/PhD_Thesis_Datasets/04_smartmeter_dataset")
raw_dir = dataset_root / "raw"
bundle_dir = dataset_root / "bundle"

raw_dir.mkdir(parents=True, exist_ok=True)
bundle_dir.mkdir(parents=True, exist_ok=True)

csv_path = raw_dir / "energydata_complete.csv"

# -----------------------------
# 4. Download (ROBUST)
# -----------------------------
def download_file(url, dest):
    headers = {"User-Agent": "Mozilla/5.0"}
    with requests.get(url, headers=headers, stream=True) as r:
        r.raise_for_status()
        with open(dest, "wb") as f:
            for chunk in r.iter_content(chunk_size=1024*1024):
                if chunk:
                    f.write(chunk)

if not csv_path.exists():
    print("\nDownloading dataset...")
    download_file(DOWNLOAD_URL, csv_path)
    print("Downloaded:", csv_path)
else:
    print("\nUsing cached file:", csv_path)

print("File size:", round(csv_path.stat().st_size/1024/1024,2), "MB")

# -----------------------------
# 5. Manifest
# -----------------------------
manifest = {
    "dataset_name": DATASET_NAME,
    "source_page": SOURCE_PAGE,
    "download_url": DOWNLOAD_URL,
    "file": str(csv_path),
    "used_in": [
        "Notebook 1: Preprocessing",
        "Notebook 3: NEO-Forecast",
        "Notebook 7: NEO-Robust",
        "Notebook 10: NEO-Edge",
        "Notebook 11: Simulation"
    ]
}

manifest_path = dataset_root / "dataset_manifest.json"
with open(manifest_path, "w") as f:
    json.dump(manifest, f, indent=2)

# -----------------------------
# 6. Create ZIP (Drive)
# -----------------------------
full_zip_path = dataset_root / "smartmeter_FULL_PACKAGE.zip"

with zipfile.ZipFile(full_zip_path, "w", zipfile.ZIP_DEFLATED) as zf:
    zf.write(csv_path, arcname="raw/energydata_complete.csv")
    zf.write(manifest_path, arcname="manifest/dataset_manifest.json")

print("ZIP created in Drive:", full_zip_path)

# -----------------------------
# 7. Copy for PC download
# -----------------------------
runtime_copy = Path("/content/smartmeter_FULL_PACKAGE.zip")
shutil.copy2(full_zip_path, runtime_copy)

# -----------------------------
# 8. Download to PC
# -----------------------------
print("\nDownloading to PC...")
files.download(str(runtime_copy))

print("\n===== SUCCESS =====")
print("Saved in Google Drive:", dataset_root)
```
